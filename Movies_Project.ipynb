## Methodology

### Methodology. Section 1: Era Classification

In this research films are categorised into four distinct cinematic eras:

1. **Golden Age of Hollywood (1927-1950):** Dominance of musicals, dramas and noir
2. **Age of Television Competition (1950-1970):** Epic films and Westerns
3. **Era of First Blockbusters (1970-2000):** Rise of Sci-Fi and action films
4. **Modern Era of Franchises (2000-2023):** Reign of superheroes and animation

**Rationale for Era Classification:**

Rather than treating cinema as a continuous 96-year period, this analysis divides the sound era into four distinct periods, each characterised by unique technological, economic and cultural conditions that fundamentally shaped filmmaking and audience preferences.

The **Golden Age (1927-1950)** represents the studio system's dominance and the establishment of sound cinema conventions. This era ended as television emerged as a competing entertainment medium (Thompson and Bordwell, 2019). The **Television Competition Era (1950-1970)** saw Hollywood respond to declining audiences through widescreen spectacles and epic productions that television could not replicate (Gomery, 2005). The **First Blockbusters Era (1970-2000)** began with *Jaws* and *Star Wars*, establishing the modern blockbuster model (Harris, 2009) with increased production budgets, merchandising and sequel strategies (Cook, 2000). Finally, the **Modern Franchise Era (2000-2023)** is defined by the dominance of shared universes, particularly superhero franchises and the rise of digital production and distribution (Pitt, 2018).

This periodisation enables meaningful comparison across structurally different industry contexts, revealing how genre preferences evolved in response to changing technological capabilities, distribution models and competitive landscapes.

### Methodology. Section 2: Genre Processing and Revenue Allocation

The TMDB dataset assigns multiple genres to most films. 
Inception (2109), for example, is classified as "Action, Science Fiction, Adventure - three genres.

**The problem:** If we count Inception's entire $1,153.56 million revenue towards each of is three genres, we would triple-count this revenue, inflating the action, science fiction, and adventure totals and misrepresenting the actual scale of the film market.

  **Available ways to solve the problem:**
- **Full attribution** (count entire revenue for each genre): Simple but inflates totals by average 2.7× (average genres/film in the data set), distorting true market size
- **Primary genre only**: Simple but loses information about genre combinations that define modern cinema
- **Equal proportional allocation:** Each film's revenue is divided equally amongst its assigned genres.
- **Weighted by position**: Assumes first genre listed is "most important," but TMDB does not rank genres by importance

**Decision: Equal proportional revenue allocation** - Each film's revenue is divided equally amongst its assigned genres.

**Example:** 
- Film: *Inception* (2010)
- Revenue (2023 USD): \$1,153.56 million
- Genres: Action, Science Fiction, Adventure (3 genres)
- Allocation per genre: \$1,153.56 ÷ 3 = \$384.52 million per genre

**Why this approach is superior:**

- **No double-counting:** Total genre revenues equal actual box office total.
- **Fairness and consistency:** Each genre contributes equally to the film's commercial success, reflecting the assumption that all elements contribute equally to audience appeal.
- **Accuracy:** Prevents inflating total cinema market size by (in case of Inception movie) 3×
- **Temporal consistency:** Total revenue by era always equals actual box office total for that era

###  Methodology. Section 3: Consumer Price Index (CPI) for inflation adjustment

**The Challenge of Temporal Comparison:** 
Comparing box office revenues across nearly a century presents a fundamental analytical challenge: the purchasing power of currency changes dramatically over time. A film earning 10 million USD in 1950 represents vastly different commercial success than one earning 10 million in 2023. Without adjustment, nominal revenue figures would systematically undervalue older films and create false impressions of genre popularity trends. For example, it is incorrect to compare the \$3 million revenue of *Gone with the Wind* (1939) with the \$3 billion of *Avatar*.

**The Solution - Consumer Price Index (CPI):** The Consumer Price Index (CPI), published by the U.S. Bureau of Labor Statistics, provides a standardised measure of price changes over time by tracking the cost of a representative basket of goods and services. By converting all revenues to 2023 equivalent values using annual CPI data, we achieve temporal comparability that reflects genuine differences in commercial success rather than monetary inflation (U.S. Bureau of Labor Statistics, 2026).

**Adjustment Formula:**

For each film, the inflation-adjusted revenue is calculated as:

**Revenue (2023 USD) = Revenue (Nominal) × (CPI₂₀₂₃ / CPIᵣₑₗₑₐₛₑ ᵧₑₐʀ)**

Where:
- Revenue (Nominal) = Original reported box office revenue
- CPI₂₀₂₃ = Consumer Price Index for 2023 (base year)
- CPIᵣₑₗₑₐₛₑ ᵧₑₐʀ = Consumer Price Index for the film's release year

**Example:**
- *Gone with the Wind* (1939) earned \$402.35 million in nominal terms
- CPI₁₉₃₉ = 13.9
- CPI₂₀₂₃ = 304.702
- Adjusted revenue = \$402.35M × (304.702 / 13.9) = **\$8,819.97M** (2023 USD)

**Justification:**

CPI adjustment is the industry-standard approach used by authoritative sources such as Box Office Mojo and The Numbers when calculating inflation-adjusted rankings (Nash Information Services, 2024). This methodology enables fair comparison of films released across different economic contexts and ensures that observed genre trends reflect genuine shifts in audience preferences rather than artifacts of monetary inflation.

#### Data Wrangling. Section 3: Millions Conversion

Budget and revenue figures in the original dataset are expressed in whole dollars (e.g., 825532764). Whilst numerically precise, such large numbers are difficult to interpret and compare visually.

**Transformation:** Budget and revenue figures will be divided by 1,000,000, expressing values in millions of dollars.

**Benefits:**
1. **Readability:** A figure like 825.53M is immediately more comprehensible than 825532764
2. **Visual clarity:** When plotted, charts become more legible and proportional relationships clearer
3. **Convention:** Industry reporting commonly uses millions as the standard unit for film financial figures

```python
# Choose base year and get its CPI
base_year = 2023
base_cpi = df_merged.loc[df_merged['year'] == base_year, 'cpi_index'].iloc[0]

# Create Temporary CPI factor to convert each year to 2023 prices
df_merged['cpi_factor'] = base_cpi / df_merged['cpi_index']

# Create two new columns: budget_real_M and revenue_real_M and adjust their figures in *millions*
df_merged['budget_real_M']  = (df_merged['budget']  * df_merged['cpi_factor']) / 1_000_000
df_merged['revenue_real_M'] = (df_merged['revenue'] * df_merged['cpi_factor']) / 1_000_000

# Round for nicer display
df_merged['budget_real_M']  = df_merged['budget_real_M'].round(2)
df_merged['revenue_real_M'] = df_merged['revenue_real_M'].round(2)

# Remove cpi_factor column
df_merged.drop(columns=['cpi_factor'], inplace=True)

# Quick check
df_merged.head(5)
```

#### Data Wrangling. Section 5: Era Classification

Rather than treating cinema as a continuous 96-year period, this analysis divides the sound era into four distinct cinematic periods, each characterised by different technological capabilities, market conditions, and audience preference.

**Four Cinematic Eras:**

1. Golden Age of Hollywood (1927–1950)
2. Age of Television Competition (1950–1970)
3. Era of First Blockbusters (1970–2000)
4. Modern Era of Franchises (2000–2023)

```python
# Define and apply era classification
def assign_era(year):
    """Assign cinematic era based on release year"""
    if 1927 <= year < 1950:
        return "Golden Age (1927-1950)"
    elif 1950 <= year < 1970:
        return "TV Competition (1950-1970)"
    elif 1970 <= year < 2000:
        return "First Blockbusters (1970-2000)"
    elif 2000 <= year <= 2023:
        return "Modern Franchises (2000-2023)"
    else:
        return "Other"

# Apply era assignment
df_merged['era'] = df_merged['year'].apply(assign_era)

# Create aggregation
era_stats = df_merged.groupby('era').agg({
    'title': 'count',
    'revenue_real_M': 'mean',
    'year': ['min', 'max']
}).round(2)
era_stats.columns = ['Film Count', 'Avg Revenue CPI Adjusted (M)', 'Start', 'End']

# Define the correct chronological order
era_order = [
    "Golden Age (1927-1950)",
    "TV Competition (1950-1970)",
    "First Blockbusters (1970-2000)",
    "Modern Franchises (2000-2023)",
    "Other"
]

# Reindex to show in chronological order
era_stats = era_stats.reindex(era_order)

# Remove "Other" if it has no data (NaN values)
era_stats = era_stats.dropna(how='all')

# print(era_stats)
#df_merged.head(5)
```

Now each film assigned to one of four eras based on release year, enabling comparative 
analysis of genre preferences across distinct historical periods.

#### Data Wrangling. Section 6: Genre Processing - Revenue Allocation Methodology

Since films in TMDB are classified with multiple genres, I employ **equal proportional revenue allocation** to ensure each genre receives its fair share of the film's revenue without double-counting or distortion.

```python
# Explode genres and allocate revenue proportionally
def explode_genres_with_allocation(dataframe):
    """
    Explode genres and allocate revenue equally across all genres per film.
    For each film:
    1. Split genres string into list
    2. Create separate row per genre
    3. Divide revenue_real_M by number of genres
    4. Divide budget_real_M by number of genres
    This ensures total revenue across all genres equals original revenue.
    """
    # Make a copy to avoid modifying original
    df_work = dataframe.copy()
    
    # First, count genres per film
    df_work['genre_count'] = df_work['genres'].str.split(', ').str.len()
    
    # Split genres into list
    df_work['genres_list'] = df_work['genres'].str.split(', ')
    
    # Explode genres - creates separate row for each genre
    df_exploded = df_work.explode('genres_list')
    
    # Clean up genre names (remove any leading/trailing whitespace)
    df_exploded['genres'] = df_exploded['genres_list'].str.strip()
    
    # Allocate revenue by dividing by genre count
    df_exploded['revenue_allocated_M'] = df_exploded['revenue_real_M'] / df_exploded['genre_count']
    df_exploded['budget_allocated_M'] = df_exploded['budget_real_M'] / df_exploded['genre_count']
    
    # Drop temporary columns
    df_exploded = df_exploded.drop(['genres_list', 'genre_count'], axis=1)
    
    # Reset index for cleaner dataframe
    df_exploded = df_exploded.reset_index(drop=True)
    
    return df_exploded

# Apply corrected genre explosion and allocation
df_genres = explode_genres_with_allocation(df_merged)

print("="*80)
print("GENRE PROCESSING RESULTS")
print("="*80)
print(f"Original dataset: {len(df_merged):,} films")
print(f"Exploded dataset: {len(df_genres):,} film-genre combinations")
print(f"Average genres per film: {len(df_genres) / len(df_merged):.2f}")

#print("\n" + "="*80)
#print("GENRE FREQUENCY")
#print("="*80)
#print(df_genres['genres'].value_counts())

# Verify allocation accuracy
print("\n" + "="*80)
print("REVENUE ALLOCATION VERIFICATION")
print("="*80)

# Check with films that have revenue
df_with_revenue = df_merged[df_merged['revenue_real_M'] > 0].copy()
df_genres_with_revenue = df_genres[df_genres['revenue_allocated_M'] > 0].copy()

total_original = df_with_revenue['revenue_real_M'].sum()
total_allocated = df_genres_with_revenue['revenue_allocated_M'].sum()

print(f"\nOriginal total revenue: ${total_original:,.2f}M")
print(f"Allocated total revenue: ${total_allocated:,.2f}M")
print(f"Difference: ${abs(total_original - total_allocated):,.2f}M")
print(f"Percentage difference: {abs(total_original - total_allocated) / total_original * 100:.6f}%")

if abs(total_original - total_allocated) < 1:  # Allow for rounding errors
    print("\n✓ ALLOCATION VERIFIED: Totals match perfectly!")
else:
    print("\n⚠ WARNING: Allocation totals do not match!")

# Show example allocation
print("\n" + "="*80)
print("EXAMPLE: Inception (2010)")
print("="*80)

inception = df_merged[df_merged['title'] == 'Inception']
if len(inception) > 0:
    print(f"\nOriginal film record:")
    print(f"  Title: {inception['title'].values[0]}")
    print(f"  Genres: {inception['genres'].values[0]}")
    print(f"  Revenue (original): ${inception['revenue_real_M'].values[0]:.2f}M")
    print(f"  Number of genres: {len(inception['genres'].values[0].split(', '))}")
    
    inception_genres = df_genres[df_genres['id'] == inception['id'].values[0]]
    print(f"\nAfter allocation (exploded into {len(inception_genres)} rows):")
    for _, row in inception_genres.iterrows():
        print(f"  {row['genres']:20s}: ${row['revenue_allocated_M']:10.2f}M")
    print(f"  Total allocated:        ${inception_genres['revenue_allocated_M'].sum():10.2f}M")
```

Each film now contributes a proportional share of its revenue to each assigned genre, ensuring analytical accuracy and preventing artificial inflation of genre revenues.

Allocated by genre revenue and also allocated by genre budget are now in columns `revenue_allocated_M` and `budget_allocated_M`, both expressed in millions.

###  Data Analysis and Visualisation. Section 6: Genre Market Share Heatmap

```python
df = df_genres.copy()

# Total allocated revenue by era and genre
genre_era_rev = (
    df.groupby(['era', 'genres'], as_index=False)['revenue_allocated_M']
      .sum()
)

# Market share (%) within each era
genre_era_rev['Market_Share_%'] = (
    genre_era_rev.groupby('era')['revenue_allocated_M']
                 .transform(lambda x: x / x.sum() * 100)
)

# Limit to top N genres overall
top_genres_overall = (
    genre_era_rev.groupby('genres')['revenue_allocated_M']
                 .sum()
                 .sort_values(ascending=False)
                 .head(15)          # keep in sync with your design
                 .index
)
genre_era_rev = genre_era_rev[genre_era_rev['genres'].isin(top_genres_overall)]

# Pivot for heatmap
heat_data = genre_era_rev.pivot(
    index='era',
    columns='genres',
    values='Market_Share_%'
).fillna(0)

era_order = [
    "Golden Age (1927-1950)",
    "TV Competition (1950-1970)",
    "First Blockbusters (1970-2000)",
    "Modern Franchises (2000-2023)"
]
heat_data = heat_data.reindex(era_order)

# Order genres by Golden Age market share (descending)
golden_age = "Golden Age (1927-1950)"
if golden_age in heat_data.index:
    cols_sorted = (
        heat_data.loc[golden_age]
        .sort_values(ascending=False)
        .index
    )
    heat_data = heat_data[cols_sorted]

# Heatmap
plt.figure(figsize=(14, 6))
sns.set(style="white")

ax = sns.heatmap(
    heat_data,
    cmap="YlGnBu",
    annot=True,
    fmt=".1f",
    cbar_kws={"label": "Market Share (%)"}
)

plt.title(
    "Genre Market Share Across Eras\n"
    "Based on CPI-Adjusted Allocated Revenue",
    fontsize=17,
    fontweight="bold"
)
plt.xlabel("Genre", fontsize=12)
plt.ylabel("Era", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)

plt.tight_layout()
plt.savefig('genre_market_share_heatmap.png', dpi=300, bbox_inches='tight')
plt.show()

print("✓ Heatmap saved: genre_market_share_heatmap.png")
```

**War and History's Decline:** War films fell from 10% to 1%, likely due to changing cultural attitudes, while History remained minimal (1-5%), suggesting audiences prefer escapism over historical realism.

### Data Analysis and Visualisation. Section 7: Individual Genre Trajectories - Small Multiples

```python
df = df_genres.copy()

# 1. Define eras in chronological order
era_order = [
    "Golden Age (1927-1950)",
    "TV Competition (1950-1970)",
    "First Blockbusters (1970-2000)",
    "Modern Franchises (2000-2023)"
]

# 2. Genres of interest and order
focus_genres = [
    "Action",
    "Adventure",
    "Drama",
    "Comedy",
    "Thriller",
    "Horror",
    "Family",
    "Romance",
    "Science Fiction"
]

# 3. Total allocated revenue by era and genre
genre_era_rev = (
    df.groupby(['era', 'genres'], as_index=False)['revenue_allocated_M']
      .sum()
)

# 4. Market share (%) within each era
genre_era_rev['Market_Share_%'] = (
    genre_era_rev.groupby('era')['revenue_allocated_M']
                 .transform(lambda x: x / x.sum() * 100)
)

# 5. Keep only the 9 focus genres
genre_era_focus = genre_era_rev[genre_era_rev['genres'].isin(focus_genres)].copy()

# 6. Ensure all era–genre combinations exist (fill missing with 0)
all_idx = pd.MultiIndex.from_product(
    [era_order, focus_genres],
    names=['era', 'genres']
)
genre_era_focus = (
    genre_era_focus
    .set_index(['era', 'genres'])
    .reindex(all_idx, fill_value=0)
    .reset_index()
)

# 7. Sort eras in chronological order
genre_era_focus['era'] = pd.Categorical(
    genre_era_focus['era'],
    categories=era_order,
    ordered=True
)
genre_era_focus = genre_era_focus.sort_values(['genres', 'era'])

# 8. Small multiples: one line chart per genre
sns.set(style="whitegrid")
fig, axes = plt.subplots(3, 3, figsize=(14, 10), sharex=True, sharey=True)
axes = axes.flatten()

for i, genre in enumerate(focus_genres):
    ax = axes[i]
    data_g = genre_era_focus[genre_era_focus['genres'] == genre]

    ax.plot(
        data_g['era'],
        data_g['Market_Share_%'],
        marker='o',
        linewidth=2
    )

    ax.set_title(genre, fontsize=11, fontweight='bold')
    ax.set_xticklabels(era_order, rotation=45, ha='right', fontsize=8)
    if i % 3 == 0:
        ax.set_ylabel('Market Share (%)')

# Hide any unused subplots (in case)
for j in range(len(focus_genres), len(axes)):
    fig.delaxes(axes[j])

fig.suptitle(
    "Individual Genre Market Share Trajectories Across Eras\n"
    "(Based on CPI-Adjusted Allocated Revenue)",
    fontsize=17,
    fontweight='bold',
    y=1.02
)
plt.tight_layout()
plt.savefig('small_multiples.png', dpi=300, bbox_inches='tight')
plt.show()

print("✓ Small Multiples saved: genre_market_share_heatmap.png")
```

**Growing Market Share:**

- **Action:** Shows the most dramatic growth trajectory, nearly vertical from 1% to 15% - the clearest winner in modern cinema
- **Adventure:** Steady upward climb from 2% to 14%, particularly accelerating in the Modern era
- **Science Fiction:** Exponential growth from near-zero to 7%, especially post-2000 with Computer-Generated Imagery (CGI) advancements

**Stable Performers:**

- **Comedy:** Relatively flat with a peak in the 1970-2000 era (16%), then slight decline to 13% - remarkably consistent
- **Thriller:** Maintains 7-9% across all eras, showing steady demand for suspense regardless of period
- **Horror:** Remains low but stable (0-3%), carving out a consistent niche market

**Declining Market Share:**

- **Drama:** Shows the steepest decline, halving from 20% to 10%
- **Romance:** Dramatic fall from 13% to 4%, one of the most significant declines
- **Family:** Downward trend from 13% to 6%, though less steep than Romance

The Small Multiples visualization tells a clear story: **cinema has fundamentally shifted from intimate, character-focused genres (Drama, Romance) toward large-scale, spectacle-driven genres (Action, Adventure, Science Fiction).** This transformation reflects:

- Globalization (action translates better internationally than dialogue-heavy drama)
- Technological capabilities (CGI enables previously impossible spectacles)
- Franchise economics (action/adventure lend themselves to sequels and universes)
- Changing audience preferences (escapism over realism)

Heatmap and Small Multiples visualizations together reveal that modern cinema is increasingly dominated by just three genres - Action, Adventure and Comedy - which collectively represent over 40% of the market, while traditional prestige genres like Drama and Romance have been marginalized.

## Conclusion

This analysis has examined the evolution of cinematic genre preferences across 96 years of sound cinema, using inflation-adjusted box office revenue from 6,429 professionally produced films in the TMDB dataset. The findings reveal substantial shifts in audience preferences, with quantifiable evidence of cinema's transformation from character-driven narratives towards spectacle-oriented entertainment.

The data demonstrates three key patterns. First, Drama's market dominance has eroded dramatically, declining from 20% of box office revenue in the Golden Age (1927-1950) to just 10% in the Modern Era (2000-2023). Second, Action has emerged as the clear commercial leader, growing from 1% to 15% market share, surpassing all other genres in total revenue ($152,993M). Third, the genre landscape has become increasingly concentrated, with Action, Adventure and Comedy collectively accounting for a disproportionate share of modern box office receipts. These shifts align with broader industry changes, including the rise of franchise filmmaking, increased production budgets and the globalisation of cinema markets (Pitt, 2018).

The methodological approach employed here - CPI adjustment, proportional revenue allocation and era-based categorisation - ensures analytical rigour and temporal comparability. However, several limitations warrant acknowledgement. The TMDB dataset, whilst comprehensive, may contain reporting inconsistencies or missing revenue data for certain films, particularly from earlier eras. The $10 million budget threshold, whilst excluding low-budget productions, may inadvertently filter culturally significant independent films that influenced genre development. Additionally, using revenue as the sole success metric overlooks critical acclaim, cultural impact and artistic innovation, which may not correlate directly with commercial performance.

This research has enhanced understanding of long-term genre dynamics through systematic empirical analysis. The findings have practical implications for industry stakeholders, suggesting that modern audiences increasingly favour high-concept, visually-driven entertainment over traditional dramatic narratives. For filmmakers and studios, these patterns indicate where commercial opportunities lie, though they should not be interpreted as prescriptive mandates that stifle creative diversity.

The recommendations for further research outlined earlier in this report identify three promising directions for extending this work. First, cross-validating TMDB financial data against alternative authoritative sources such as Box Office Mojo and The Numbers would strengthen confidence in findings and reveal potential systematic differences in genre categorisation and revenue reporting. Second, investigating geographic variations in genre performance across regions (North America, Europe, Asia-Pacific) would illuminate whether the identified trends reflect genuine global shifts or are driven primarily by specific markets. Third, analysing how streaming platforms have affected theatrical genre distribution since 2010 would provide insight into how digital disruption has reshaped genre economics and audience consumption patterns. These avenues would deepen understanding of the complex interplay between technology, culture and commercial success in contemporary cinema.
